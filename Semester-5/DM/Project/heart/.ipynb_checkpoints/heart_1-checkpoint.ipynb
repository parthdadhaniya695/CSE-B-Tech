```python
# Import necesary Modules
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load the dataset
df = pd.read_csv("heart.csv")
df
```

```python
# 1. What are the first 5 rows of the dataset?
df.head()
```

```python
# 2. How many rows and columns are there in the dataset?
df.shape
```

```python
# 3. What are the column names?
df.columns.tolist()
```

```python
# 4. What are the data types of each column?
df.dtypes
```

```python
# 5. Are there any missing values in the dataset?
df.isnull().sum()
```

```python
# 6. What is the summary (mean, std, min, max, etc.) of each numeric column?
df.describe()
```

```python
# 7. How many unique values are there in each column?
df.nunique()
```

```python
# 8. How many patients have heart disease (target=1) vs not (target=0)?
df['target'].value_counts()
```

```python
# 9. What is the percentage of patients with heart disease?
df['target'].value_counts(normalize=True) * 100
```

```python
# 10. Plot a bar chart of heart disease vs no disease.
df['target'].value_counts().plot(kind='bar', color=['skyblue', 'salmon'])
plt.title('Heart Disease Count')
plt.xlabel('Target')
plt.ylabel('Count')
plt.xticks([0, 1], ['No Disease', 'Disease'], rotation=0)
plt.show()
```

```python
# 11. What is the average age of patients?
avg_age = df['age'].mean()
print(avg_age)
```

```python
# 12. What is the distribution of gender (sex)?
df['sex'].value_counts()
```

```python
# 13. What is the average age for men and women separately?
df.groupby('sex')['age'].mean()
```

```python
# 14. Which age group has the most heart disease cases?
df[df['target'] == 1]['age'].value_counts().sort_values(ascending=False).head()
```

```python
# 15. Count of chest pain types (cp)?
df['cp'].value_counts()
```

```python
# 16. What is the maximum and minimum resting blood pressure (trestbps)?
df['trestbps'].max(), df['trestbps'].min()
```

```python
# 17. What is the average cholesterol level (chol)?
avg_chol = df['chol'].mean()
print(avg_chol)
```

```python
# 18. Show correlation matrix between numeric columns.
df.corr(numeric_only=True)
```

```python
# 19. Which features are most correlated with heart disease (target)?
df.corr(numeric_only=True)['target'].sort_values(ascending=False)
```

```python
# 20. Plot histogram of age.
df['age'].hist(bins=20, color='lightgreen')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
# 21. Boxplot of cholesterol levels for patients with and without heart disease.
df.boxplot(column='chol', by='target', grid=False)
plt.title('Cholesterol by Heart Disease')
plt.suptitle('')
plt.xlabel('Heart Disease')
plt.ylabel('Cholesterol')
plt.show()
```

```python
# 22. What is the average maximum heart rate (thalach) for people with and without heart disease?
df.groupby('target')['thalach'].mean()
```

```python
# 23. What is the count of fasting blood sugar > 120 (fbs=1)?
df['fbs'].value_counts()
```

```python
# 24. Which combination of sex and chest pain type (cp) is most common?
df.groupby(['sex', 'cp']).size().sort_values(ascending=False).head()
```

```python
# 25. Show average values of all features grouped by heart disease target.
df.groupby('target').mean(numeric_only=True)
```
